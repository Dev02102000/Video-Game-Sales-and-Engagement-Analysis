```python
def clean_title(title):
    if pd.isnull(title):
        return ''
    return str(title).strip().lower().replace(":", "").replace("-", "").replace(",", " ").replace(".", "")

games['clean_title'] = games['Title'].apply(clean_title)
sales['clean_name'] = sales['Name'].apply(clean_title)

merged = pd.merge(games, sales, how= 'left', left_on='clean_title', right_on='clean_name')

merged.drop(columns=['clean_title', 'clean_name'], inplace=True)

merged.to_csv("merged_games.csv", index=False)

print("MERGING COMPLETED. final shape:", merged.shape)
```

```python
cursor.execute("USE game_salesdata_db")
cursor.execute("""
CREATE TABLE IF NOT EXISTS games_data (
    unnamed INT,
    title VARCHAR(255),
    release_date VARCHAR(20),
    team VARCHAR(100),
    rating VARCHAR(10),
    times_listed VARCHAR(50),
    number_of_reviews VARCHAR(50),
    genres VARCHAR(255),
    summary TEXT,
    reviews TEXT,
    plays VARCHAR(50),
    playing INT,
    backlogs INT,
    wishlist INT
)
""")
conn.commit()
print("table created successfully.")
```

```python
# select only the columns that match the games_data table (drop the extra `clean_title` column)
cols = [
    'Unnamed: 0', 'Title', 'Release Date', 'Team', 'Rating',
    'Times Listed', 'Number of Reviews', 'Genres', 'Summary',
    'Reviews', 'Plays', 'Playing', 'Backlogs', 'Wishlist'
]

query = """
    INSERT INTO games_data (unnamed, title, release_date, team, rating, times_listed, number_of_reviews, genres, summary, reviews, plays, playing, backlogs, wishlist)
    VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

def parse_count(x):
    """Parse strings like '3.8K', '2M', '230' into integers (or None)."""
    if pd.isnull(x):
        return None
    s = str(x).strip().lower().replace(',', '')
    try:
        if s.endswith('k'):
            return int(float(s[:-1]) * 1000)
        if s.endswith('m'):
            return int(float(s[:-1]) * 1_000_000)
        return int(float(s))
    except Exception:
        return None

# work on a copy to avoid modifying original dataframe
df_trim = games[cols].copy()

# convert the integer columns to actual ints (these were defined as INT in MySQL)
for c in ['Playing', 'Backlogs', 'Wishlist']:
    if c in df_trim.columns:
        df_trim[c] = df_trim[c].apply(parse_count)

# replace NaN with None so MySQL receives NULL for missing values
data_list_trimmed = df_trim.where(pd.notnull(df_trim), None).values.tolist()

cursor.executemany(query, data_list_trimmed)
conn.commit()
print("data inserted using to_list")
```
